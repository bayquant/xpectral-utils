# RATE LIMITER

Throttle calls to an API with `RateLimiter`: at most `calls` acquisitions per `per_seconds` window.

```python
%load_ext autoreload
%autoreload 2
```

```python
# Standard library imports
import time

# First-party imports
from xpectral_utils.rate_limiter import RateLimiter

CALLS = 3
PER_SECONDS = 1
```

## Throttle a loop

Call `acquire()` before each request. The first `CALLS` go through immediately; the next one blocks until the oldest call leaves the window.

```python
limiter = RateLimiter(calls=CALLS, per_seconds=PER_SECONDS)
start = time.monotonic()

for i in range(9):
    limiter.acquire()
    print(f"call {i} at {time.monotonic() - start:.1f}s")
```

## Invalid arguments

`calls` and `per_seconds` must be positive.

```python
try:
    RateLimiter(calls=0, per_seconds=1)
except ValueError as error:
    print(error)
```
